```python
import os
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt 
import matplotlib.gridspec as gridspec
import scipy 

from matplotlib.patches import Ellipse

import seaborn as sns
from IPython.display import Markdown as md
```

```python
# compute the influence function of log-density 
def inffun_logden_normal_loc(x, mean, y): 
    
    output = (x - mean) * (y - mean)
    
    return output
```

```python
# compute the influence function of log-density 
def inffun_den_normal_loc(x, mean, y): 
    
    output = np.exp(-(x - mean) ** 2 / 2) / np.sqrt(2 * np.pi) * (x - mean) * (y - mean)
    
    return output
```

```python
mean = 0.
y = 2.
x = np.linspace(-5., 5., 2000)

inffun_logden_vals = inffun_logden_normal_loc(x = x, mean = mean, y = y)
inffun_den_vals = inffun_den_normal_loc(x = x, mean = mean, y = y)
```

```python
label_fontsize = 20
info_fontsize = 35
tick_fontsize = 15
legend_fontsize = 20
linewidth = 5.0
x_label = r'evaluation point'
y_logden_label = r'$\mathrm{IF}_x (T, F, 2)$'
y_den_label    = r'$\mathrm{IF}_x (\widetilde{T}, F, 2)$'

den_plot_xlim = (31., 290.)
logden_plot_ylim = (-775., 1099.)
den_plot_ylim = (-0.035, 0.028)

contam_data = 250.
legend_loc = 'lower right'

nrows, ncols = 1, 2
fig = plt.figure(figsize=(21, 10))
gs = gridspec.GridSpec(nrows = nrows, ncols = ncols)

ax1 = fig.add_subplot(gs[0, 0])
ax1.plot(x, inffun_logden_vals, linewidth = linewidth, color = 'tab:blue')

ax1.axhline(0., 0, 1, ls = '--', color = 'tab:purple', alpha = 0.8)
ax1.axvline(0., 0, 1, ls = '--', color = 'tab:purple', alpha = 0.8)
ax1.axvline(y, 0, 1, ls = (0, (5, 10)), color = 'k', alpha = 0.8)
# ax1.set_xlim(den_plot_xlim)
# ax1.set_ylim(logden_plot_ylim)
ax1.set_xlabel(x_label, fontsize = label_fontsize)
ax1.set_ylabel(y_logden_label, fontsize = label_fontsize)
ax1.tick_params(axis = 'both', labelsize = tick_fontsize)

ax2 = fig.add_subplot(gs[0, 1])
ax2.plot(x, inffun_den_vals, linewidth = linewidth, color = 'tab:blue')

ax2.axhline(0., 0, 1, ls = '--', color = 'tab:purple', alpha = 0.8)
ax2.axvline(0., 0, 1, ls = '--', color = 'tab:purple', alpha = 0.8)
ax2.axvline(y, 0, 1, ls = (0, (5, 10)), color = 'k', alpha = 0.8)
# ax2.set_xlim(den_plot_xlim)
# ax2.set_ylim(den_plot_ylim)
ax2.set_xlabel(x_label, fontsize = label_fontsize)
ax2.set_ylabel(y_den_label, fontsize = label_fontsize)
ax2.tick_params(axis = 'both', labelsize = tick_fontsize)

plt.savefig(f'../plots/comparison-inffunden-inffunlogden-normal-loc-model-mean={mean}-y={y}.pdf', 
            bbox_inches = 'tight')
```
